```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer, util
import torch
import requests
```

```python
path = '/home/baga/Downloads/статусы_по_аналитическим_системам_МИО_и_ЦГО.xlsx'
```

```python
df = pd.read_excel(path)
```

```python
df
```

```python
dev = torch.device('cpu')
model = SentenceTransformer('LaBSE').to(dev)
```

```python
df.columns
```

```python
corpus_title = df["Название "].tolist()
corpus_description = df["описание"].tolist()
```

```python
def concat(x: str, y: str) -> str:
    x = x if not pd.isnull(x) else ''
    y = y if not pd.isnull(y) else ''
    return x + ' ' + y
```

```python
corpus = [concat(x, y) for x, y in zip(corpus_title, corpus_description)]
```

```python
embedded_corpus = model.encode(corpus, convert_to_tensor=True)
```

```python
embedded_corpus
```

```python
torch.save(embedded_corpus, 'embedded_corpus_v2.pt')
```

```python
class ModelLoader:
    def __init__(self):
        self.embedded_corpuses = torch.load('/home/baga/Desktop/ML_MODEL/embedded_corpus.pt', map_location=torch.device('cpu'))
        self.dataset = pd.read_excel(path)
        self.dev = torch.device('cpu')
        self.model = SentenceTransformer('LaBSE').to(self.dev)

    def search(self, query_corpus: str):
        question_embedding = self.model.encode(query_corpus, convert_to_tensor=True)
        hits = util.semantic_search(question_embedding, self.embedded_corpuses)[0]

        corpus_ids = [hit['corpus_id'] for hit in hits[:10]]

        result_dataset = self.dataset[self.dataset.index.isin(corpus_ids)].reindex(corpus_ids).reset_index(drop=True)

        return result_dataset[['Название ', 'ссылка']].to_dict(orient='records')
```

```python
engine = ModelLoader()
```

```python
for i in engine.search('категория семей')[:3]:
    html_link = f'''<a href="{i['ссылка']}">{i['Название '].strip()}</a>'''
    print(html_link)
```

```python
def matched_result(key_words:str):
    engine = ModelLoader()
    for i in engine.search(key_words)[:3]:
        html_link = f'''<a href="{i['ссылка']}">{i['Название '].strip()}</a>'''
        print(html_link)
```
